```python
import io

file_name = list(uploaded.keys())[0]

excel = pd.ExcelFile(io.BytesIO(uploaded[file_name]))

print("Available Sheets:")
print(excel.sheet_names)
```

```python
for sheet in excel.sheet_names:
    print("\nSHEET:", sheet)

    df_temp = pd.read_excel(excel, sheet_name=sheet)

    print("Columns:")
    print(df_temp.columns.tolist())

    display(df_temp.head())
```

```python
df = pd.read_excel(
    io.BytesIO(uploaded[file_name]),
    sheet_name=excel.sheet_names[0]
)

# Remove duplicate records
df = df.drop_duplicates()

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check dataset information
print("\nDataset Information:")
df.info()

# Display dataset
display(df.head(10))
```

```python
import io
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the '🏥 Visit' sheet which contains diagnostic/condition data
visit_df = pd.read_excel(io.BytesIO(uploaded[file_name]), sheet_name='🏥 Visit')

# Clean column names to prevent key errors
visit_df.columns = visit_df.columns.str.strip()

# 2. Extract and compute the top 5 diagnosed conditions
# 'Diagnosis' or similar column contains the medical condition. Let's find the correct column name dynamically.
diagnosis_col = [col for col in visit_df.columns if 'diagnos' in col.lower() or 'condition' in col.lower()][0]
top_5_conditions = visit_df[diagnosis_col].value_counts().head(5)

# 3. Plot the bar chart
plt.figure(figsize=(10, 5))
sns.barplot(
    x=top_5_conditions.values,
    y=top_5_conditions.index
)

plt.title("Top 5 Diagnosed Conditions")
plt.xlabel("Number of Cases")
plt.ylabel("Diagnosis")

plt.tight_layout()
plt.show()
```

```python
import io
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the '👤 Patient' sheet which contains demographics including Age
patient_df = pd.read_excel(io.BytesIO(uploaded[file_name]), sheet_name='👤 Patient')

# Clean column names
patient_df.columns = patient_df.columns.str.strip()

# 2. Plot the patient age distribution
plt.figure(figsize=(9, 5))
sns.histplot(patient_df['Age'].dropna(), bins=15, kde=True)

plt.title("Patient Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")

plt.tight_layout()
plt.show()
```

```python
import io
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load the '🏥 Visit' sheet if it's not already loaded (already in visit_df)
# Clean columns to avoid key errors
visit_df.columns = visit_df.columns.str.strip()

# 2. Find the column name representing "Follow-Up"
followup_col = [col for col in visit_df.columns if 'follow' in col.lower()][0]

# 3. Compute follow-up counts
followup_counts = visit_df[followup_col].value_counts()

# 4. Plot the pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    followup_counts,
    labels=followup_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Patient Follow-Up Analysis")
plt.show()
```

```python
cost_data = df.groupby('Diagnosis')['TreatmentCost'].mean()
cost_data = cost_data.sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=cost_data.values,
    y=cost_data.index
)

plt.title("Average Treatment Cost by Diagnosis")
plt.xlabel("Average Treatment Cost")
plt.ylabel("Diagnosis")

plt.tight_layout()
plt.show()
```

```python
visit_data = (
    df.groupby('Diagnosis')['VisitID']
    .nunique()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=visit_data.values,
    y=visit_data.index
)

plt.title("Patient Visits by Diagnosis")
plt.xlabel("Number of Visits")
plt.ylabel("Diagnosis")

plt.tight_layout()
plt.show()
```
